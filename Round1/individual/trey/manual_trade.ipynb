```python
import numpy as np
```

```python
class Currencies:
    snowballs = "SNOWBALLS"
    pizzas = "PIZZAS"
    silicon_nuggets = "SILICON_NUGGETS"
    seashells = "SEASHELLS"
    
conversion_rates = {Currencies.snowballs:{Currencies.pizzas:1.45,Currencies.silicon_nuggets:0.52,Currencies.seashells:0.72},
                    Currencies.pizzas:{Currencies.snowballs:0.7,Currencies.silicon_nuggets:0.31,Currencies.seashells:0.48},
                    Currencies.silicon_nuggets:{Currencies.snowballs:1.95,Currencies.pizzas:3.1,Currencies.seashells:1.49},
                    Currencies.seashells:{Currencies.snowballs:1.34,Currencies.pizzas:1.98,Currencies.silicon_nuggets:0.64}}
```

```python
def calculate(path : list) -> float:
    num = 1
    for i in range(len(path)-1):
        start = path[i]
        end = path[i+1]
        # print(start,end)
        num *= conversion_rates[start][end]
    return num
```

```python
# Two edges
paths_2 = []
profit = []
for currency in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
    path = []
    path.append(Currencies.seashells)
    path.append(currency)
    path.append(Currencies.seashells)
    paths_2.append(path)
    profit.append(calculate(path))
print(max(profit))
print(paths_2[np.argmax(profit)])
```

```python
# Three edges
paths_3 = []
profit = []
for currency1 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
    for currency2 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
        if currency2!=currency1:
            path = []
            path.append(Currencies.seashells)
            path.append(currency1)
            path.append(currency2)
            path.append(Currencies.seashells)
            paths_3.append(path)
            profit.append(calculate(path))
print(max(profit))
print(paths_3[np.argmax(profit)])
```

```python
# Four edges
paths_4 = []
profit = []
for currency1 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
    for currency2 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets, Currencies.seashells]:
        for currency3 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
            if currency2!=currency1:
                if currency3!=currency2:
                    path = []
                    path.append(Currencies.seashells)
                    path.append(currency1)
                    path.append(currency2)
                    path.append(currency3)
                    path.append(Currencies.seashells)
                    paths_4.append(path)
                    profit.append(calculate(path))
print(max(profit))
print(paths_4[np.argmax(profit)])
```

```python
# Five edges
paths_5 = []
profit = []
for currency1 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
    for currency2 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets, Currencies.seashells]:
        for currency3 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets,Currencies.seashells]:
            for currency4 in [Currencies.snowballs, Currencies.pizzas, Currencies.silicon_nuggets]:
                if currency2!=currency1:
                    if currency3!=currency2:
                        if currency4!=currency3:
                            path = []
                            path.append(Currencies.seashells)
                            path.append(currency1)
                            path.append(currency2)
                            path.append(currency3)
                            path.append(currency4)
                            path.append(Currencies.seashells)
                            paths_5.append(path)
                            profit.append(calculate(path))
print(max(profit))
print(paths_5[np.argmax(profit)])
```
